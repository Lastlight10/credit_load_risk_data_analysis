##### 2. Grade C Loans have the most Defaulted Loans

The analyst examined the defaulted loans and discovered that most of the defaulted loans have the Grade C. This was done using the following code:

```python
import pandas as pd # Import dependency

dataset = pd.read_csv('cleaned_credit_risk_dataset.csv') # Import dataset
defaulted_loans = dataset.loc[dataset['Clean_cb_person_default_on_file'] == True, 'Clean_loan_grade'] # Get defaulted loans

# Get all count for each grade
A_grades = defaulted_loans[defaulted_loans == 'A']
B_grades = defaulted_loans[defaulted_loans == 'B']
C_grades = defaulted_loans[defaulted_loans == 'C']
D_grades = defaulted_loans[defaulted_loans == 'D']
E_grades = defaulted_loans[defaulted_loans == 'E']
F_grades = defaulted_loans[defaulted_loans == 'F']
G_grades = defaulted_loans[defaulted_loans == 'G']

# Create the dataframe table
defaulted_loans_table = pd.DataFrame(
  [A_grades.count(),B_grades.count(),C_grades.count(),D_grades.count(),E_grades.count(),F_grades.count(),G_grades.count()],
  index=['A','B','C','D','E','F','G'],
  columns=['Count']
)
defaulted_loans_table
```

The results above show the number of defaulted loans and their grades. Grade C has the most defaulted loans even though the grade is considered to only be at the 'Average Risk' compared to the lower grades. The data analyst recommends the company the following points regarding this insight:

  * Deeper Analysis of Defaulted Loans
    * The grade in which the defaulted loans are most prominent is Grade C, which is at average risk. The grade signifies how likely a loan can be completely paid where the lower the grade, the more risk there is for a loan to default. However, the results show that most defaulted loans are not from higher risk grades, but from the loans with average risk. This findings can be caused by wrong grade assignment because of insufficient analysis of the borrower's background. To investigate this further, the analyst recommends to conduct more examination and review of the defaulted loans and the data that are related to them.

The dates are important for understanding the credit risk invloved in the loans. They can determine how long loans has been active, if the dates support seasonal trends or patterns, or if the loan is completed earlier than the expected date of completion. The analysis also needs the dates to determine if the loan is recent or outdated and dates can also be used as categories when filtering and analyzing data from specific date periods. The data analsyt recommends the following points:

  * Include Dates for Loan Disbursement, Date of Defaulting and Loan Completion
    * As mentioned before, the dates can be used to provide additional information and context for loans. In data analysis, these dates are beneficial in identifying trends or seasonal patterns. The data analyst recommends for the current and future datasets to include dates when the loan was officialy given to the client, the date when the loan was considered to be defaulted, the expected date of completion and when the loan was actually fully paid.
