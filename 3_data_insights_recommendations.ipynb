### Data Insights and recommendations

In this part of the notebook, the data analyst shared additional findings that are relevant to the dataset and gave the aproppriate recommendations for the company. The insights and recommendations are listed below:

#### 1. Few Loans with High Amount

The analysis of the dataset revealed loans where the amount is high but they number only a few, making them outliers to the rest of the borrowers. While not necessarily a bad thing, it is beneficial for a company to attract more borrowers with higher loan amounts than the average. The data analyst suggested the following points:

  * Marketing Campaigns, Promos and Special Deals
    * To attract borrowers, the company can make special deals or promos for clients that want to borrow money at a higher amount. Since from the analysis, there are loan amounts that can reach up to almost $40,000 with a median of $8,000.
