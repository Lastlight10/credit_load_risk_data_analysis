---
### Table Description

The following table showcases the columns involved and their descriptions.

| Column Name | Description |
| --- | --- |
| person_age | Age of the loaner |
| person_income | Annual income |
| person_home_ownership | Status of home ownership |
| person_emp_length | Employment length in terms of years |
| loan_intent | Reason of loaning |
| loan_grade | Grade associated with the loaner |
| loan_amnt | Amount of money to loan |
| loan_int_rate | Interest rate |
| loan_status | Status of the loan. ( 1 means the loan is still active) |
| loan_percent_income | Percentage of income that goes into the loan. |
| cb_person_default_on_file | If the loaner has failed to make scheduled payments|
| cb_person_cred_hist_length | Length of credit history in years |



#### Currency

Due to the dataset not having a default currency, Philippine Peso (₱) will be used on the data analysis.

#### Loan Grade

There are many ways a loan can be graded. In the dataset, it uses an alphabetical grading instead ranging from A to F.
