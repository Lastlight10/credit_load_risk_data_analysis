```python
person_ages[person_ages != 144].max() # Get the new maximum age
```

Excluding `144`, another issue has occured in which the new maximum value that is not likely to occur considering human lifespan, there is also a chance that this is an entry error. Since the new possible maximum age is `94`, there is a very high chance that this is the new maximum value. Now that we have a new valid maximum age, let us deal with the outliers by identifying their other details.

To treat values with `144`, the analyst have considered the following options as the new value:

> `null`: the analyst have considered to use null as the replacement value in order to not skew the statistics but having considered that all the columns are needed to be analyzed carefully, `null` will not be a suitable value in this case.

> `14`: Considering that the age is too low for a loan, and will standout as the only value lower then the minimum value found, `14` is not a possible value. The income also cannot be viable if the person is 14.

> `44`: The only possible value because it is in the range between minimum and maximum values and at the range of ages where loans are still applicable to apply.

To treat values with `123`, the analyst have considered the following options as the new value:

> `null`: the analyst have also considered to use null as the replacement value in order to not skew the statistics but having considered that all the columns are needed to be analyzed carefully, `null` will not be a suitable value in this case.

> `12 or 13`: These ages also considered too low for a loan, therefore not a suitable value. The income also cannot be viable if the person is 12 or 13.

> `21 or 31`: Also possible but very unlikely since the keyboard layout which starts from 1 then 2 and 3. If it is `21` or `31`, we are more likely to see `121` or `231`.

> `23`: The only possible value because it is also in the range between minimum and maximum values and at the range of ages where loans are still applicable to apply.

Based on the consideration and data review, it is more applicable to replace `144` with `44` and `123` with `23`. In order to change the values, the following code is used:

##### Loan Interest Rate

This column represents the interest rate for each loan per year. Interest rate is important for credit loaners as it is how businesses earn their profit. To further examine the data, the following code was used:

```python
interest_rates = dataset['loan_int_rate']
interest_rates.info()
```

```python
interest_rates.describe()
```

The output shows that only `29465` out of the total `32581` entries have values. This column is important to be filled correctly because if the missing values are simply filled carelessly, it can skew the output. To examine the problem further, the analyst calculated the percentage of null values using the following code:

```python
(interest_rates.isnull().sum() /  interest_rates.count()) * 100 # Find the percentage of null values
```

The result showed that almost `11%` of the loan interest rate data is null and randomly filling out these values will definitely skew the output. One way to solve this issue is to use MICE or Multivariate Imputation by Chained Equations. This is a technique used to handle missing data by filling it with values based on the relationship of data with other columns. To achieve this, we use the following module from `sklearn`:

```python
null_rates = interest_rates[interest_rates.isnull()]
null_rates # Get the rows of null interest rates
```

```python
from sklearn.experimental import enable_iterative_imputer  
from sklearn.impute import IterativeImputer

#  Dataset with missing loan values and columns to base the data from.
data = {
    'person_income': cleaned_dataset['Clean_person_income'],
    'loan_amnt': cleaned_dataset['Clean_loan_amnt'],
    'interest_rate': dataset['loan_int_rate']
}
df = pd.DataFrame(data) # Store relevant information in this dataframe

# Initialize the MICE Imputer
imputer = IterativeImputer(max_iter=12, random_state=40)
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)
interest_rates = df_imputed['interest_rate']
```

```python
df_imputed['interest_rate'].describe()
```

After the MICE imputation, the previously nulled values were filled varying rates making the total entries of rows that has no null values to be `32581` rows. The next process involves the following code:

```python
cleaned_dataset = cleaned_dataset.drop(columns=['loan_int_rate'])
cleaned_dataset['Clean_loan_int_rate'] = interest_rates
cleaned_dataset['Clean_loan_int_rate'].to_csv('cleaned_data/cleaned_loan_int_rate.csv', index=False)
cleaned_dataset
```
