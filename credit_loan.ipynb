##### Age

This column represents the ages of the loaners when the dataset was recorded. Analyzing this column can provide insights regarding the ages and other data. To further examine the age of the loaners, the analyst have used the following code that targets the age of the loaners specifically by assigning the column into a separate series of data:

```python
person_ages = dataset['person_age']
person_ages.info()
```

The result shows the data as a series of integers consisting of 32,581 entries.

```python
person_ages.describe()
```

Using `.describe()` shows the statistical summary of the column. According to the function results, the maximum value is `144` which should be impossible for a human lifespan and treated as a data entry error. To further analyze the concern, the analyst finds the ages equivalent to `144` and the maximum age lower than the value.

```python
person_ages[person_ages != 144].max()
```

Excluding `144`, another issue has occured in which the new maximum value that is not likely to occur considering human lifespan, there is also a chance that this is an entry error. That new number is `123` which is close to the oldest age a person has achieved which is 122. To analyze this further, let us examine the remaining data without the values of `123` and `144`.

```python
person_ages[(person_ages != 144) & (person_ages != 123)].max()
```

Since the new possible maximum age is `94`, there is a very high chance that this is the new maximum value. This is because `94` is a valid age to get a loan like [PVAO Pension Loans](https://www.veteransbank.com.ph/pension-loan) which allows citizens 65 years old and above to qualify for a row. Now that we have a new valid maximum age, let us deal with the outliers by identifying their other details.

```python
filtered_ages = dataset[['person_age','person_income']]
filtered_ages[ (filtered_ages['person_age'] == 144) | ( filtered_ages['person_age']  == 123)]
```

```python
person_ages[person_ages.eq(123)].count()
```

```python
person_ages[person_ages.eq(144)].count()
```

The results indicate that the rows with `144` as the age only amount up to three rows:`81`, `183`, `32297`. On the other hand, only two rows with the value of `123` are found: `575`, `747`.

To treat values with `144`, the analyst have considered the following options as the new value:

> `null`: the analyst have considered to use null as the replacement value in order to not skew the statistics but having considered that all the columns are needed to be analyzed carefully, `null` will not be a suitable value in this case.

> `14`: Considering that the age is too low for a loan, and will standout as the only value lower then the minimum value found, `14` is not a possible value. The income also cannot be viable if the person is 14.

> `44`: The only possible value because it is in the range between minimum and maximum values and at the range of ages where loans are still applicable to apply.

To treat values with `123`, the analyst have considered the following options as the new value:

> `null`: the analyst have also considered to use null as the replacement value in order to not skew the statistics but having considered that all the columns are needed to be analyzed carefully, `null` will not be a suitable value in this case.

> `12 or 13`: These ages also considered too low for a loan, therefore not a suitable value. The income also cannot be viable if the person is 12 or 13.

> `21 or 31`: Also possible but very unlikely since the keyboard layout which starts from 1 then 2 and 3. If it is `21` or `31`, we are more likely to see `121` or `231`.

> `23`: The only possible value because it is also in the range between minimum and maximum values and at the range of ages where loans are still applicable to apply.

Based on the consideration and data review, it is more applicable to replace `144` with `44` and `123` with `23`. In order to change the values, the following code is used:

```python
age_with_144 = filtered_ages['person_age'] == 144
filtered_ages.loc[age_with_144, 'person_age'] = 44
filtered_ages.iloc[[81,183,32297]]
```

```python
age_with_144 = filtered_ages['person_age'] == 123
filtered_ages.loc[age_with_144, 'person_age'] = 23
filtered_ages.iloc[[575,747]]
```

During the process, the analyst stored the rows with the age values of `123` and `144` and change the value of `'person_age'` for the said rows to match the chosen value which is `23` for `123` and `44` for `144`. Then the same rows were examined again for their new values and based on the result, the ages has changed, therefore it can be concluded that the column has been cleaned and the same changes can now be used on the main copy of the dataset with cleaned data. The cleaned column is also stored as a separate CSV file in the `cleaned_data` folder for backup. This can be done using the following code:

```python
cleaned_dataset = dataset.copy()
cleaned_dataset = cleaned_dataset.drop(columns=['person_age'])
cleaned_dataset['Clean_person_age'] = filtered_ages['person_age']

cleaned_dataset['Clean_person_age'].to_csv('cleaned_data/cleaned_person_ages.csv', index=False)

cleaned_dataset
```

```python
cleaned_dataset.iloc[[575,747]]
```

To wrap things up, the older column was replaced dropped and new values for their ages are not stored. The filtered column is added at the end of the table. When we take a look back to the columns with dirty values, they are now cleaned.

##### Income

This column stores the data regarding the yearly income of the loaners when the dataset was made. Income can give many insights and is one of the most important variable when considering to provide a loan. The analyst used the following code to further examine the column's data.

```python
person_incomes = dataset['person_income']
person_incomes
```

```python
person_incomes.info()
```

```python
person_incomes.describe().apply(lambda x: format(x, 'f'))
```

The python code resulted in more detailed information regarding the column. From the data, it consists of 32581 entries and as intergers with no null values. The numerical summaries are also presented like the count, mean, standard deviation, minimum and maximum value, quartiles, and maximum values. One outlier that pops up is the maximum income which amounts to `6000000` and compared to the third quartile, a considerable jump in income has been observed. To further analyze the high amount of incomes, the analyst examined all the income between the third quartile and maximum income using the following code:

```python
dataset[dataset['person_income'] == 6000000]
```

```python
third_quartile_incomes = person_incomes[(person_incomes >= person_incomes.quantile(0.75)) & (person_incomes < 6000000) ]
third_quartile_incomes.describe().apply(lambda x: format(x,'f'))
```

After gathering all the values of between the third quartile and the `6000000` income, the new maximum income is `2039784`, which means that there is a huge gap between the largest income and the second most largest income. This is important to know as now we can treat `6000000` as the only outlier in its range. Therefore, it can be surmised that there is one value in the dataset in this column that can possibly an entry error. In this case, analyzing the rows of income with the same range values as the outlier can provide insights. The code below was used:

```python
same_home_ownership=(dataset['person_home_ownership']) == 'MORTGAGE'
same_age_range =(dataset['person_age'] == 144) | (dataset['person_age'] == 44 )

x = dataset[((same_age_range) | (same_home_ownership)) & (dataset['person_income'] != 6000000 )]
x['person_income'].describe().apply(lambda x: format(x,'f'))
```

The results shows that the outlier with the income of `6000000` is truly unique compared to other loaners of the same age range

#### Loan Grade

There are many ways a loan can be graded. In the dataset, it uses an alphabetical grading, ranging from A to F based on the propriety risk rating scale. The details of the loan grades is shown below:

| Grade | Meaning |
| --- | --- |
| A | Lowest risk |
| B | Low risk |
| C | Average risk |
| D | Moderate risk |
| E | High risk |
| F | Very High risk |

The grades determine the 'risk' or the frequency of the debter to fall into a historical default. Therefore loaners tend to charge high interest rate to compensate for the possible risk.
